```python
# replace blank / NA / N/A / unknown with NaN
df = df.replace(bad, np.nan)
df.isnull().sum()
```

```python
# standardize channel names
df["Channel_Name"] = df["Channel_Name"].str.lower().str.replace(" ", "")

names = {"suntv": "Sun TV", "vijaytv": "Vijay TV", "zeetamil": "Zee Tamil", "colorstamil": "Colors Tamil",
         "kalaignartv": "Kalaignar TV", "jayatv": "Jaya TV", "rajtv": "Raj TV", "ktv": "KTV",
         "starsports": "Star Sports", "sonysports": "Sony Sports"}
df["Channel_Name"] = df["Channel_Name"].map(names)
print(df["Channel_Name"].value_counts(dropna=False))
```

```python
# only KTV is in the Movies category, so missing names there are KTV
df.loc[df["Channel_Name"].isnull() & (df["Channel_Category"] == "Movies"), "Channel_Name"] = "KTV"

# the rest cannot be found out
df["Channel_Name"] = df["Channel_Name"].fillna("Unknown")
```

```python
# Sports channels are English, others are Tamil
df.loc[df["Language"].isnull() & (df["Channel_Category"] == "Sports"), "Language"] = "English"
df["Language"] = df["Language"].fillna("Tamil")

df["Popular_Show"] = df["Popular_Show"].fillna("Unknown")
```

```python
# convert text numbers to numeric
for col in text_cols:
    df[col] = pd.to_numeric(df[col].str.replace(",", ""), errors="coerce")
print(df.dtypes)
```

```python
# invalid values -> NaN
df.loc[(df["Avg_Rating"] < 0) | (df["Avg_Rating"] > 5), "Avg_Rating"] = np.nan
df.loc[df["Subscription_Fee"] < 0, "Subscription_Fee"] = np.nan
df.loc[(df["Years_Active"] < 1) | (df["Years_Active"] > 40), "Years_Active"] = np.nan
```

```python
# fill missing numbers with median
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
print("Duplicate records:", df.duplicated().sum())
```

```python
df["Avg_Rating"] = df["Avg_Rating"].round(1)
for col in ["Avg_Viewers", "Total_Likes", "Social_Media_Followers", "YouTube_Views", "Monthly_Reach", "Years_Active", "Subscription_Fee"]:
    df[col] = df[col].round().astype(int)
```

```python
# check again
print(df.isnull().sum().sum())
print(df["Avg_Rating"].min(), df["Avg_Rating"].max())
print(df["Subscription_Fee"].min())
print(df["Years_Active"].min(), df["Years_Active"].max())
print(df["HD_Available"].unique())
print(df["Channel_Name"].unique())
df.info()
```
