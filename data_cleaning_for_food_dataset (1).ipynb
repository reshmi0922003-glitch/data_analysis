DATA CLEANING

```python
df.isnull().sum()
```

```python
df['region']=df['region'].fillna('Unknown')
print(df['region'].isnull().sum())
```

```python
df.isnull().sum()
```

```python
duplicates_count=df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates_count}")
```

```python
print(df["state"].unique())
print(df["region"].unique())
print(df["prep_time"].unique())
print(df["cook_time"].unique())
print(df["ingredients"].unique())
print(df["flavor_profile"].unique())
```

```python
df["state"]=df["state"].replace("-1", "Unknown")
```

```python
df["region"]=df["region"].replace("-1", "Unknown")
df["prep_time"]=df["prep_time"].replace("-1", df["cook_time"].mean())
df["cook_time"]=df["cook_time"].replace("-1",df["cook_time"].mean() )
df["ingredients"]=df["ingredients"].replace("-1", "Unknown")
df["flavor_profile"]=df["flavor_profile"].replace("-1", "Unknown")
```

DATA ANALYSIS

```python
df["prep_time"]=pd.to_numeric(df["prep_time"],errors="coerce")
df["cook_time"]=pd.to_numeric(df["cook_time"],errors="coerce")
```

DATA VISUALIZATION

```python
df["course"].value_counts().plot(kind="bar")
plt.title("Distribution of Course")
plt.xlabel("Course")
plt.ylabel("Count")
plt.show()
```

```python
df["ingredients"].value_counts().head(10).plot(kind="bar",color='purple')
plt.title("Distribution of Ingredients")
plt.xlabel("Ingredients")
plt.ylabel("Count")
plt.show()
```

```python
df["flavor_profile"].value_counts().head(10).plot(kind="bar",color='orange')
plt.title("Distribution of Flavor Profile")
plt.xlabel("Flavor Profile")
plt.ylabel("Count")
plt.show()
```

```python
plt.pie(df.groupby('flavor_profile')['flavor_profile'].count(), labels=df.groupby('flavor_profile')['flavor_profile'].sum().index, autopct='%1.1f%%')
plt.title('Flavor Profile Distribution')
```

```python
plt.scatter(df['prep_time'], df['cook_time'])
plt.title('Prep Time vs Cook Time')
plt.xlabel('Prep Time')
plt.ylabel('Cook Time')
```

```python
plt.plot(df['prep_time'], df['cook_time'])
plt.title('Prep Time vs Cook Time')
plt.xlabel('Prep Time')
plt.ylabel('Cook Time')
```

```python
plt.hist(df['prep_time'], bins=20, color='green', alpha=0.7)
plt.title('Distribution of Prep Time')
plt.xlabel('Prep Time')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.boxplot(df['cook_time'])
plt.title('Boxplot of Cook Time')
plt.ylabel('Cook Time')
plt.show()
```
